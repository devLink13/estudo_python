APRENDENDO A UTILIZAR A BIBLIOTECA NUMPY

--> Principais Características do NumPy:
    Arrays N-dimensionais: Suporte para arrays de qualquer dimensão, permitindo operações complexas de forma simples e eficiente.
    Funções Matemáticas: Inclui funções para álgebra linear, transformadas de Fourier, geração de números aleatórios, entre outras.
    Desempenho: O núcleo do NumPy é escrito em C, proporcionando alta performance.
    Interoperabilidade: Funciona bem com outras bibliotecas de computação científica e de dados, como SciPy, Pandas, e Matplotlib

```python
import numpy as np
#importando a biblioteca numpy
```

```python
# array tradicional python
arr = [0, 2, 3, 5]
print(arr)
print(arr[0])
```

```python
# array usando numpy
# repare que usando o numpy o array printando tem algumas diferenças na separação dos elementos
arr2 = np.array([0, 2, 3, 5])
print(arr2)
print(arr2[0])
```

```python
# a função arange gera uma PA de valor inicial como 1° arg, valor final com 2° arg, e o passo como 3° arg, além do tipo de informação que ele
# retorna, por exemplo, inteiros

arr = np.arange(0, 10, 1, int)
print(arr)

arr = np.arange(0, 50, 2, float)
print(arr)

arr = np.arange(0, 100, 1, int)
print(arr)

```

```python
# função reshape() serve para reestruturar um array sem alterar seu conteúdo

# faça um reshape para ordenar o array com 10 linhas e 10 colunas
arr = np.reshape(arr,(10,10))
print(arr)

print()

arr = np.reshape(arr,(5,20))
print(arr)

arr_mod = list()
arr_mod = np.reshape(arr, (20,5))
print(arr_mod)
```

```python
# a indexação na biblioteca numpy é semelhante a indexação padrão do python, usando colchetes, a diferença é que aqui,
# para matrizes 2d (linha, coluna) nós passamos as coordenadas linha/coluna, veja:

print(arr_mod[2,3])
print(arr_mod[1,1])
```

```python
# operação com matrizes, soma posição com posição

arr1 = np.array([ [2,3], [3,5] ])
arr2 = np.array([ [3,4], [9,8] ])

matriz_soma = arr1 + arr2
print(matriz_soma)
```

```python
#np.sum() soma todos os valores da matriz

arr = np.sum(matriz_soma)
print(arr)
```

```python
# np.zeros() gera uma matriz de 3x3 com zeros, util para iniciar uma matriz

arr = np.zeros((3,3), int)
print(arr)

print()

arr = np.zeros(5, int)
print(arr)
```

```python
#np.random.rand(), função que gera numeros aleatório entre 0 e 1, e coloca tudo isso dentro de uma matriz

arr = np.random.rand(3,3)*100
print(arr)

```

```python
# np.append() funciona exatamente igual o append() do python

arr = np.random.rand(5)*100
print(arr)

arr = np.append(arr, 35)
print(arr)
```
