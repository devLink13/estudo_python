FUNÇÕES ESTATÍSTICAS DO PANDAS

A biblioteca Pandas oferece várias funções estatísticas úteis para análise de dados. Aqui estão algumas das principais:

1. **mean()**: Calcula a média dos valores.
   ```python
   df['coluna'].mean()
   ```

2. **median()**: Determina a mediana dos valores.
   ```python
   df['coluna'].median()
   ```

3. **std()**: Calcula o desvio padrão.
   ```python
   df['coluna'].std()
   ```

4. **var()**: Calcula a variância.
   ```python
   df['coluna'].var()
   ```

5. **min() e max()**: Encontram o valor mínimo e máximo, respectivamente.
   ```python
   df['coluna'].min()
   df['coluna'].max()
   ```

6. **describe()**: Fornece um resumo estatístico, incluindo contagem, média, desvio padrão, valores mínimos e máximos, e percentis.
   ```python
   df.describe()
   ```

7. **corr()**: Calcula a correlação entre colunas.
   ```python
   df.corr()
   ```

8. **count()**: Conta o número de valores não nulos.
   ```python
   df['coluna'].count()
   ```

Essas funções permitem uma análise detalhada e eficiente dos dados numéricos³⁴. Se precisar de mais detalhes ou exemplos específicos, sinta-se à vontade para perguntar!
