```python
class Veiculo:
    registro_veicular = 540021
    proprietario = 'LOCALIZA'

    def __init__(self, marca, modelo, ano, ligado = False):
        self.marca = marca
        self.modelo = modelo
        self.ano = ano
        self.status_funciomento = ligado

    # função para tradução do estado de funcionamento do veículo
    def estado_de_funcionamento(self):
        if self.status_funciomento == False:
            self.funcionando = 'DESLIGADO'
            return self.funcionando
        
        elif self.status_funciomento == True:
            self.funcionando = 'LIGADO'
            return self.funcionando


    def ligar(self):
        if self.estado_de_funcionamento() == 'DESLIGADO':

            self.status_funciomento = True
            return f'O veículo {self.marca, self.modelo} está ligado.'
        
        else:
            return f'IMPOSSÍVEL, veículo {self.marca, self.modelo} já está ligado.'

    def desligar(self):
        if self.estado_de_funcionamento() == 'LIGADO':

            self.status_funciomento = False
            return f'O veículo {self.marca, self.modelo} está desligado.'
        
        else:
            return f'IMPOSSÍVEL, veículo {self.marca, self.modelo} já está desligado.'



## usando as variáveis no escopo de classe
print(Veiculo.registro_veicular)
print(Veiculo.proprietario)

# alterando estado de uma variável de instancia
carro = Veiculo('peugeot', '408', '2012')
carro_2 = Veiculo('peugeot','2008', '2018')

print(carro.status_funciomento)
print(carro_2.status_funciomento)

carro.ligar() #mandando carro ligar
print(carro.status_funciomento) # este será true
print(carro_2.status_funciomento) # este será false

#mandar o carro ligar de novo para receber o erro.
print(carro.ligar())
carro.desligar()

print(carro_2.ligar())
print(carro_2.status_funciomento)

print(carro_2.desligar())
print(carro_2.desligar())

# iniciar uma instancia de veículo com estado ligado
carro_3 = Veiculo('peugeot', '3008', '2015', ligado=True)
print(carro_3.ligar())
```

```python
## MÉTODOS DE CLASSE --> @classmethod

## podemos usar o classmethod para criar fabricas, que basicamente é um método que criar objetos(instancias) da própria classe
# NÃO PODEMOS USAR ATRIBUTOS DE INSTANCIA DENTRO DAS CLASSMETHODS

import random 

class Pessoa():
    
    ano_atual = 2024 #variavel de classe ou atributo de classe

    #método construtor do objeto/instancia
    def __init__(self, nome, idade, altura = None, peso = None, cor_olhos = None):
        self.nome = nome
        self.idade = idade
        self.altura = altura
        self.peso = peso
        self.cor_olhos = cor_olhos

    @classmethod
    def metodo_de_classe(cls): # é necessário passar cls para métodos de classe, assim como usamos self para métodos de instancia
        print('isso é um método de classe.')
    
    #exemplo de factory method, criar uma pessoa padrão, com os atributos padrões, que podem ser configurados depois
    # essa pessoa virá "pré configurado com o necessário, podendo alterar os atributos depois"
    @classmethod
    def criar_pessoa_padrao(cls, nome, idade):
        nome = nome
        idade = idade
        altura = 1.80
        peso = 80
        cor_olhos = 'azul'

        return Pessoa(nome, idade, altura, peso,cor_olhos)

    # imagine que seja necessário criar um menor de idade, podemos chamar um factory method para isso
    # com idade aleatoria
    @classmethod
    def criar_menor_idade(cls, nome):
        nome = nome
        idade = random.randint(12, 17)

        return Pessoa(nome, idade)


    def metodo_de_instancia(self):
        print('isso é um método de instancia')

# para chamar o método de instancia precisamos definir uma instancia da classe
pessoa_1 = Pessoa('link', 24)
pessoa_1.metodo_de_instancia()
Pessoa('wesley', 30).metodo_de_instancia() # assim tb é possível chamar o método de instancia

#um método de classe podemos chamar sem necessariamente criar uma instancia da classe.
Pessoa.metodo_de_classe()

#criando pessoa padrão
pessoa_padrao = Pessoa.criar_pessoa_padrao('AMERICO', 55)
print(pessoa_padrao.nome)
print(pessoa_padrao.idade)
print(pessoa_padrao.cor_olhos)
print(pessoa_padrao.altura)
print(pessoa_padrao.peso)

print('-'*30)

pessoa_padrao2 = Pessoa.criar_pessoa_padrao('JULIO', 25)
print(pessoa_padrao2.nome)
print(pessoa_padrao2.idade)
print(pessoa_padrao2.cor_olhos)
print(pessoa_padrao2.altura)
# altero somento o parâmetro que preciso
pessoa_padrao2.peso = 125
print(pessoa_padrao2.peso)

print('-'*30)

menor_idade = Pessoa.criar_menor_idade('Junior')
print(menor_idade.nome, menor_idade.idade)





    
```

```python
## MÉTODOS DE CLASSE --> @classmethod

# São métodos estáticos dentro da classe que não possuem acesso nem ao cls e nem ao self
# são funções que existem dentro da classe e podem ser usadas para algum fim

class Pessoa():

    ano = None # iniciando a varivável como None

    # faremos uma função estática que configurará o atributo de classe. Podemos fazer configurações na classe através deste método.
    @staticmethod
    def set_ano(ano=None):
        Pessoa.ano = ano

    #método construtor do objeto/instancia
    def __init__(self, nome, idade, altura = None, peso = None, cor_olhos = None):
        self.nome = nome
        self.idade = idade
        self.altura = altura
        self.peso = peso
        self.cor_olhos = cor_olhos

print(Pessoa.ano)
Pessoa.set_ano(2025)
print(Pessoa.ano)

# podemos criar muitas instancias e mesmo assim o estado da variavel não alterará
pessoa_1 = Pessoa('LINK', 24)
print(Pessoa.ano)
pessoa_2 = Pessoa('laura', 27)
print(Pessoa.ano)


Pessoa.set_ano(2030)
print(Pessoa.ano)
```

```python
class Connection:
    def __init__(self, host = 'localhost') -> None:
        self.host = host
        self.user = None
        self.password = None

    # método setter
    def setUser(self, user):
        self.user = user

    def setPassword(self, password):
        self.password = password

    #criando um forma de fazer uma conexão mesmo sem configurar o init
    @classmethod
    def connect_auth (cls, user, password):
        connection = cls()
        connection.user = user
        connection.password = password

        return connection

    #classe para retornar uma mensagem de log que não necessitou nem de self e nem de cls
    @staticmethod
    def msg_log(msg):
        msg_log = msg
        return msg_log

c1 = Connection()
print(c1.user)
c1.setUser('Wesley')
c1.setPassword('12345')
print(c1.user, c1.password)

c2 = Connection.connect_auth('wesley link', '122345')
print(c2.user)
print(c2.password)

```

```python
# USANDO PROPERTY COMO SETTER
# setter é setar algo, configurar algo.
# getter é obter algo, obter um valor.

## CONVENÇÃO --> ATRIBUTOS QUE COMEÇAM COM 1 OU 2 '_' NÃO DEVEM SER USADOS FORA DA CLASSE.


class Caneta:
    def __init__(self, cor) -> None:
        # private ou protect
        self._cor = cor #_cor indica que esse atributo é privado da classe, e não deve ser usado fora
        self._corTampa = None # inicia a cor da tampa como none e ela será setada pelo setter

    

    
    #QUANDO TEMOS UMA PROPERTY SE COMPORTANDO COM GETTER/SETTER NÓS PODEMOS TER DUAS FUNÇÕES DE MESMO NOME PARA REALIZAR COISAS DIFERENTES
    @property
    def cor(self):
        return self._cor
    # para configurar o setter, precisamos usar o decorator '@' e chamar a função acima com .setter ai abaixo podemos usar o nome que quisermos
    # poderiamos chamar o getter de getCor e o setter de setCor
    @cor.setter
    def cor(self, valor):
        self._cor = valor
        return None


    ########### EXEMPLO PRÁTICO DO QUE FALEI ACIMA ###########
    # @property
    # def getCor(self):
    #     print('estou no getter')
    #     return self._cor
    
    # # para configurar o setter, precisamos usar o decorator '@' e chamar a função acima com .setter ai abaixo podemos usar o nome que quisermos
    # # poderiamos chamar o getter de getCor e o setter de setCor
    # @getCor.setter
    # def setCor(self, valor):
    #     print('estou no setter',valor)


    ####### VAMOS CONFIGURAR OUTRA PROPERTY DE GETTER/SETTER PARA PEGAR E ATRIBUIR A COR DA TAMPA ##########
    @property
    def getCorTampa(self):
        if self._corTampa is None:
            return 'Cor da tampa ainda não foi definida, use setCorTampa para defini-lá.'
        else:
            return self._corTampa
    
    @getCorTampa.setter
    def setCorTampa(self, cor_tampa):
        self._corTampa = cor_tampa
        return None
    

caneta = Caneta('azul')
print(caneta.cor)

caneta2 = Caneta('preta')
print(caneta2.cor)
caneta2.cor = 'VERMELHA'
print(caneta2.cor)

caneta3 = Caneta('Verde')
print(caneta3.getCorTampa)
caneta3.setCorTampa = 'violeta'
print(caneta3.getCorTampa)
```

```python
# ENCAPSULAMENTO DE CLASSES EM PYTHON (MODIFICADORES DE ACESSO: PUBLIC, PROTECT E PRIVATE)

# PYTHON NÃO TEM MODIFICADORES DE ACESSO

# PUBLIC = PODE SER ACESSADO EM QUALQUER LUGAR (DENTRO E FORA DA CLASE)
# PROTECTED = SÓ PODE USAR EM CLASSE OU SUBCLASSE, FORA NÃO
# PRIVATE = SÓ PODE SER USADO DENTRO DA CLASSE QUE FOI CRIADO, NEM MESMO EM SUBCLASSES DEVE SER USADO

# EXISTEM CONVENÇÕES :
        #(SEM UNDERLINE) = PUBLIC (USADO EM QUALQUE LUGAR)
        #(UM UNDERLINE) = PROTECTED (SÓ DEVE SER ACESSADO DENTRO DA CLASSE E SUAS SUBCLASSES)
        #(DOIS UNDERLINE) = PRIVATE (SÓ PARA A CLASSE EM QUE ELE FOI DEFINIDO, NÃO DEVE SER USADO EM NENHUM OUTRO LOCAL)

class Foo:
    def __init__(self) -> None:
        self.public = 'isso é publico' #atributo publico
        self._protected = 'isso é protegido' #atributo protegido
        self.__private = 'isso é privado' #atributo privado

    # método publico
    def metodo_publico(self):
        print('isso é um método publico')

    # método protegido
    def _metodo_protegido(self):
        print('isso é um método protegido')

    # método privado
    def __metodo_privado(self):
        print('isso é um método privado')
```

```python
# RELAÇÃO ENTRE OBJETOS (ASSOCIAÇÃO, AGREGAÇÃO E COMPOSIÇÃO)

# UML É MUITO USADA NESSES CASOS --> LINGUAGEM DE DIAGRAMAS



######################### ASSOCIAÇÃO ##########################
class Escritor:
    def __init__(self, nome) -> None:
        self.nome = nome
        self._ferramenta = None

    @property
    def ferramenta(self):
        return self._ferramenta
    
    @ferramenta.setter
    def ferramenta(self, ferramenta):
        self._ferramenta = ferramenta


class FerramentaDeEscrever:
    def __init__(self, nome) -> None:
        self.nome = nome

    def escrever(self):
        return f'O {self.nome} está escrevendo.'
    
escritor = Escritor('luiz')
caneta = FerramentaDeEscrever('Caneta Bic')
maquina_de_escrever = FerramentaDeEscrever('Máquina')
escritor.ferramenta = caneta
print(caneta.escrever())
print(escritor.ferramenta.escrever())
escritor.ferramenta = maquina_de_escrever
print(escritor.ferramenta.escrever())
```

```python
########################### AGREGAÇÃO #############################
## AGREGAÇÃO AINDA É UMA UNIÃO FRACA DE DUAS CLASSES, ONDE AMBAS AS CLASSES EXISTEM SOZINHAS, PORÉM JUNTAS CONSEGUEM FUNCIONAR MELHOR
# PERCEBA QUE A CLASSE PRODUTOS SÓ EXISTE PARA ARMAZENAR OS PRODUTOS

# UM CARRINHO DE COMPRAS AGREGA UM OU MAIS PRODUTOS #

class Carrinho:
    def __init__(self):
        #lista de produtos
        self._produtos = []
        self._precoProdutos = []

    def inserir_produtos(self, *produtos):
        for produto in produtos:
            self._produtos.append(produto)

    def listar_produtos(self):
        for produto in self._produtos:
            print(f'PRODUTO: {produto.nome}, PREÇO: {produto.preço}')
    
    def calcular_preco_total (self):
        for produto in self._produtos:
            self._precoProdutos.append(produto.preço)
            preço_total = sum(self._precoProdutos)
        return print('R$',preço_total)




class Produtos:
    def __init__(self, nome, preço):
        self.nome = nome
        self.preço = preço


# criar um carrinho
carrinho = Carrinho()

# criar produtos
p1 = Produtos('banana', 1.50)
p2 = Produtos('abacaxi', 3.50)

# inserir os produtos no carrinho, AQUI É FEITA A LIGAÇÃO DAS DUAS CLASSES, PERCEBA: AO PASSAR COMO PARÂMETRO OS DOIS PRODUTOS QUE SÃO INSTANCIAS DE PRODUTOS
# NÓS CRIAMOS UMA AGREGAÇÃO.
carrinho.inserir_produtos(p1,p2)
# listar os produtos que estão no carrinho
carrinho.listar_produtos()

carrinho.calcular_preco_total()
```

```python
############### HERANÇA ################
# Herança simples - Relações entre classes
# Associação - usa, Agregação - tem
# Composição - É dono de, Herança - É um
#
# Herança vs Composição
#
# Classe principal (Pessoa)
#   -> super class, base class, parent class
# Classes filhas (Cliente)
#   -> sub class, child class, derived class


# AFINAL PARA QUE SERVE HERANÇA? PARA TORNAR UMA CLASSE MAIS GENÉRICA, USANDO ATRIBUTOS DE OUTRA CLASSE

# POR EXEMPLO: UMA PESSOA PODE SER MUITAS COISAS, COMO POR EXEMPLO SER UM PROGRAMADOR, UM POLICIAL, UM MILITAR, UM ENGENHEIRO,
# LOGO CADA UMA DESSAS PROFISSÕES PODERIAM HERDAM ATRIBUTOS DE PESSOA, UMA VEZ QUE TODO PROGRAMADOR, POLICIAL, MILITAR OU ENGENHEIRO TAMBÉM SÃO PESSOAS

######## OBS IMPORTANTE: AO USAR HERANÇA USE COM MODERAÇÃO, POIS UMA HERANÇA DE MUITOS NÍVEIS, NIVEIS DO METHOD RESOLUTION ORDER, PODE TORNAR O CÓD. MT COMPLEXO



# essa é a superclasse
class Pessoa:
    cpf = 1234

    def __init__(self, nome, sobrenome):
        self.nome = nome
        self.sobrenome = sobrenome

    def falar_nome_classe(self):
        #falará o nome do objeto que está sendo e executado e falará o nome da classe que está sendo executada no momento
        print(self.nome, self.sobrenome, self.__class__.__name__)    

class Policial(Pessoa):
       # OBS: FOI CLONADA A FUNÇÃO falar_nome_classe(self) justamente para entender o method resolution order, o interpretador encontrará o método
       # dentro da classe Policial primeiro, logo não irá procurar na classe pai, ou ainda, classe Pessoa.
       def falar_nome_classe(self):
            #falará o nome do objeto que está sendo e executado e falará o nome da classe que está sendo executada no momento
            print('não sai da classe Policial')
            print(self.nome, self.sobrenome, self.__class__.__name__)   

class Militar(Pessoa):
    pass

pessoa_wesley = Policial('wesley', 'link')
pessoa_wesley.falar_nome_classe()

pessoa_maria = Militar('maria', 'antonieta')
pessoa_maria.falar_nome_classe()

# veja que o atributo de classe da classe Pessoa é comum a ambos
print(pessoa_maria.cpf)
print(pessoa_wesley.cpf)

# usar o help() para ver as informações importantes da classe, como o method resolution order e etc..
# method resolution order informa a ordem na qual o interpretador python buscará as informações dentro da classe, superclasse e objeto
#help(pessoa_wesley)
```

```python
########### SUPER E SOBREPOSIÇÃO DE MÉTODOS #############

"""
 super() é uma classe que faz referência à classe a qual herdamos algo.

"""

class A:
    atributo_a = 'valor a'

    def metodo(self):
        print('A')

# observe que class B herda tudo de A, porém o método 'metodo' será sobreposto, fazendo com que ao chamar 'metodo' execute apenas o da classe B.
class B(A):
    atributo_b = 'valor b' 

    def metodo(self):
        print('=-'*30)
        print('B')

        # após executar o método de b, iremos executar o método de a
        #super(B, self).metodo() # podemos usar assim
        super().metodo() #ou podemos usar assim
        print('=-'*30)

# agora por sua vez, C herda de B, porém o método 'metodo' será sobreposto em B, o que acarretará apenas na execucação do método em B.
class C(B):
    atributo_c = 'valor c'
    

    def metodo(self):
        # repare que abaixo eu configurei as supers para que ao chamar o método que tem o mesmo nome em todas classes ele não seja sobreposto
        # ou ainda, mesmo que ele seja sobreposto, ele executará os outros métodos das outras classes.

        print('=-'*30)
        print('C')
        # veja abaixo o método explicito de chamar a super de B, não é necessário pagar B,self como parâmetro para referenciar a super de B.
        # ao usar .metodo() ao lado super, estamos dizendo o seguinte: Interpretador, vai lá na minha super (nesse caso a classe B) e executa o método chamado 'metodo'
        #super(B, self).metodo()
        super().metodo() # receberemos  B

        # leia o seguinte: interpretador, vai na classe A (que é super de B), e apartir dela procure o método 'metodo' e execute ele.
        super(B, self).metodo() # receberemos A
        print('=-'*30)

a = A()
b = B()
c = C()
# observe que C herdou os atributos de b e a, mesmo que eles não tenham sido definidos em c. Isso é herança.
print(c.atributo_c, c.atributo_b, c.atributo_a)

# no momento que eu chamar c.metodo() ele irá executar o método de c, não o de B e muito menos o de A.
c.metodo()
print()
b.metodo()
print()
a.metodo()
# se eu quiser que ele execute o método da superclasse a qual ele herda, eu preciso usar a classe super().
```

```python
################## HERANÇA MULTIPLA - UMA CLASSE HERDA DE DUAS OU MAIS #################################
## CLASS CARRO(FABRICANTE, MOTOR)

'''
    --> uma herança simples (aquela que tende a buscar a generalização de algo) se for muito profunda pode ser muito complexo de programar.
    --> herança multipla, se mal estruturada, sem conhecimento do mro (method resolution order) pode ser muito mais complexo
'''
# Herança Múltipla - Python Orientado a Objetos
# Quer dizer que no Python, uma classe pode estender
# várias outras classes.
# Herança simples:
# Animal -> Mamifero -> Humano -> Pessoa -> Cliente
# Herança múltipla e mixins
# Log -> FileLog
# Animal -> Mamifero -> Humano -> Pessoa -> Cliente
# Cliente(Pessoa, FileLog)
#
# A, B, C, D
# D(B, C) - C(A) - B(A) - A
#
# método -> falar
#           A
#         /   \
#        B     C
#         \   /
#           D
#
# Python 3 usa C3 superclass linearization
# para gerar o mro.
# Você não precisa estudar isso (é complexo)
# https://en.wikipedia.org/wiki/C3_linearization
#
# Para saber a ordem de chamada dos métodos
# Use o método de classe Classe.mro()
# Ou o atributo __mro__ (Dunder - Double Underscore)

                                        #busca na D        DEPOIS NA B              DEPOIS NA C         DEPOIS NA A            OBJECT
# esse é o mro da classe abaixo: [<class '__main__.D'>, <class '__main__.B'>, <class '__main__.C'>, <class '__main__.A'>, <class 'object'>]
# VALE LEMBRAR -------> ISSO SÓ ACONTECE PORQUE OS MÉTODOS ESTÃO SENDO SOBREPOSTOS (MÉTODOS TÊM MESMO NOME EM CLASSES DIFERENTES QUE SE HERDAM)

class A:
    ...

    def quem_sou(self):
        print('A')


class B(A):
    ...

    # def quem_sou(self):
    #     print('B')


class C(A):
    ...

    def quem_sou(self):
        print('C')


class D(B, C):
    ...

    def quem_sou(self):
        print('D')


d = D()
d.quem_sou()
# print(D.__mro__)
print(D.mro())
```
