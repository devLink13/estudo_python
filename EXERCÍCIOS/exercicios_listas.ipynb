5.Faça um Programa que peça as quatro notas de 10 alunos, calcule e armazene num vetor a média de cada aluno, imprima o número de alunos com média maior ou igual a 7.0.

        - gerar 4 notas aleatórias do tipo float de 0 a 10, fazer isso para cada aluno.
        - calcular a media de cada aluno e armazenar num vetor de médias
        - imprimir o numero de alunos com média maior ou igual a 7.0
        

```python
import random

# CONSTANTES
N_ALUNOS = 10
N_AVALIACOES = 4
MEDIA_CORTE = 7

lista_alunos = list()
lista_notas = list()
for aluno in range(N_ALUNOS):
    #print('aluno', aluno)
    
    for nota in range(N_AVALIACOES):
        nota = random.random() * 10
        nota = round(nota, 3)
        #print('\t', nota)
        lista_notas.append(nota)
    lista_auxiliar = lista_notas.copy() # gerar uma cópia da lista de 4 notas
    #print(f'{lista_auxiliar=}')
    lista_alunos.append(lista_auxiliar) # após gerar lista com 4 notas, adicionar a lista principal
    lista_notas.clear() # limpa a lista de notas

#print(lista_alunos[0])

lista_medias = list()

for aluno, notas in enumerate(lista_alunos):
    media = sum(notas) / len(notas)
    media = round(media, 3)
    #print(aluno , media)
    lista_medias.insert(aluno, media)

for aluno, media in enumerate(lista_medias):
    print(f'O aluno {aluno + 1}, obteve as seguintes notas {lista_alunos[aluno]}, ficando com média {media}.')


num_medias_maiorq_7 = 0
for media in lista_medias:
    num_medias_maiorq_7 += 1 if media >= MEDIA_CORTE else 0

percentual = (num_medias_maiorq_7 / len(lista_medias)) * 100 # percentual de aprovados


print(f'{num_medias_maiorq_7} alunos obtiveram nota maior que 7, representando {percentual:.1f}% de aprovados.')


  

```

8. Foram anotadas as idades e alturas de 30 alunos. Faça um Programa que determine quantos alunos com mais de 13 anos possuem altura inferior à média de altura desses alunos.

```python
import random

lista_geral = list()

for aluno in range(10):
    idade = random.randint(10, 30)
    altura = random.uniform(1.5, 2)
    lista_aluno = [f'aluno {aluno+1}',idade, altura]
    lista_aluno_copia = lista_aluno.copy()
    lista_geral.append(lista_aluno_copia)
    lista_aluno.clear()
    #print('\t', idade, altura)

#print(lista_geral)

media_altura = 0
for aluno in lista_geral:
    media_altura += aluno[2]

media_altura = media_altura / len(lista_geral)
media_altura = round(media_altura, 2)

lista_mais_altos = list()
for aluno in lista_geral:
    if aluno[1] > 13:
        if aluno[2] > media_altura:
            lista_mais_altos.append(aluno[0])
qtd_mais_altos = len(lista_mais_altos)

print(f'A média de altura dos alunos é {media_altura}, ao todo são {qtd_mais_altos} de alunos maiores de 13 anos mais altos que a média.')
```

9. Faça um programa que receba a temperatura média de cada mês do ano e armazene-as em uma lista. Após isto, calcule a média anual das temperaturas e mostre todas as temperaturas acima da média anual, e em que mês elas ocorreram (mostrar o mês por extenso: 1 – Janeiro, 2 – Fevereiro, . . . ).

```python
import random

lista_temperaturas_mensais = list()
for mes in range(1, 13):
    temperatura = round(random.uniform(25, 40), 2)
    lista_mes = [mes, temperatura] # lista para cada mes contendo mes e temp media
    lista_aux = lista_mes[:]
    lista_temperaturas_mensais.append(lista_aux)
    lista_mes.clear()
print(lista_temperaturas_mensais)

media_anual = 0
for mes in lista_temperaturas_mensais:
    media_anual = media_anual + mes[1]

media_anual = round((media_anual/12), 2)
# print(media_anual)

lista_meses_mais_quentes = list()
for mes in lista_temperaturas_mensais:
    if mes[1] > media_anual:
        lista_meses_mais_quentes.append(f'mês {mes[0]}, temperatura {mes[1]}')

print(f'a média de temperaturas anual foi de {media_anual}')
print(lista_meses_mais_quentes)




```

Utilize uma lista para resolver o problema a seguir. Uma empresa paga seus vendedores com base em comissões. O vendedor recebe $200 por semana mais 9 por cento de suas vendas brutas daquela semana. Por exemplo, um vendedor que teve vendas brutas de $3000 em uma semana recebe $200 mais 9 por cento de $3000, ou seja, um total de $470. Escreva um programa (usando um array de contadores) que determine quantos vendedores receberam salários nos seguintes intervalos de valores:
        $200 - $299
        $300 - $399
        $400 - $499
        $500 - $599
        $600 - $699
        $700 - $799
        $800 - $899
        $900 - $999
        $1000 em diante
Desafio: Crie ma fórmula para chegar na posição da lista a partir do salário, sem fazer vários ifs aninhados.

```python
#exercicio interessante... precisamos 'casar' intervalos de duas listas, para isso usamos uma formula
#em vez de usar ifs aninhados


import random
intervalos = [0] * 9
print(intervalos)

def calculo_salarial(vendas_brutas):
    return round(200 + (0.09 * vendas_brutas), 2)

vendas = []
for c in range(10):
    venda = random.randint(0, 12000)
    vendas.append(venda)

print(vendas)
qtd_vendas = len(vendas)

intervalos_salariais = [
    '$200 - $299',
    '$300 - $399',
    '$400 - $499',
    '$500 - $599',
    '$600 - $699',
    '$700 - $799',
    '$800 - $899',
    '$900 - $999',
    '$1000 em diante',
]

for venda in vendas:
    salario = calculo_salarial(venda)
    print(salario)
    if salario >= 1000:
        intervalos[8] += 1
    else:
        posicao = (int(salario) - 200) // 100 # essa formula indicará a posicao dentro da faixa salarial
        intervalos[posicao] += 1

for indice, salario in enumerate(intervalos_salariais):
    print(f'Intervalo de {salario}, vendedores {intervalos[indice]}')
```

Em uma competição de salto em distância cada atleta tem direito a cinco saltos. O resultado do atleta será determinado pela média dos cinco valores restantes. Você deve fazer um programa que receba o nome e as cinco distâncias alcançadas pelo atleta em seus saltos e depois informe o nome, os saltos e a média dos saltos. O programa deve ser encerrado quando não for informado o nome do atleta. A saída do programa deve ser conforme o exemplo abaixo:
        
        Atleta: Rodrigo Curvêllo
 
        Primeiro Salto: 6.5 m
        Segundo Salto: 6.1 m
        Terceiro Salto: 6.2 m
        Quarto Salto: 5.4 m
        Quinto Salto: 5.3 m

        Resultado final:
        Atleta: Rodrigo Curvêllo
        Saltos: 6.5 - 6.1 - 6.2 - 5.4 - 5.3
        Média dos saltos: 5.9 m


```python
lista_informacoes_atleta = [0] * 6
lista_memoria =  list()
while True:
    nome_atleta = input('Atleta: ')
    lista_informacoes_atleta[0] = nome_atleta

    if nome_atleta:
        for c in range(1, 6):
            salto = input(f'{c}° salto: ')
            lista_informacoes_atleta[c] = float(salto)
        
        media_saltos = sum(lista_informacoes_atleta[1:]) / 5
        
        print('RESULTADO FINAL:')
        print(f'Atleta: {lista_informacoes_atleta[0]}')
        print(f'Saltos: {str(lista_informacoes_atleta[1:]).replace('[', ''). replace(']', '').replace(',', ' - ')}')
        print(f'Media dos saltos: {media_saltos} m')

        # INCREMENTAR UMA MEMÓRIA - COM LISTA
        lista_informacoes_atleta.append(media_saltos)
        lista_aux = lista_informacoes_atleta[:] # cria uma cópia
        lista_memoria.append(lista_aux)

    else:
        break # acabo o programa aqui

print(lista_memoria)
```
