# INTRODUÇÃO AOS DICIONÁRIOS ' DICT '

São estruturas de dados do tipo par de chave valor. Isto é, uma chave é única de cada valor. As chaves podem ser consideradas como “índices” pois como vimos em listas precisamos acessar um conjunto de dados por um índice (na lista os índices são numéricos, nos dicionários podem ser qualquer tipo de dados imutáveis, isto é, int, float, bool, str ou tuple) já o valor do par pode ser de qualquer tipo inclusive outro dicionário. Vale a pena lembrar que dict e list são dados mutáveis que podem ser modificados. 

Exemplo de um dicionário em python:

```python
pessoa = dict() # forma de declarar um dicionário  vazio
pessoa = dict(nome = 'wesley', sobrenome='link') # podemos criar pares usando dict desta forma

#a forma mais comum de criarmos dicionários é usado {} (chaves)
# lembre que aqui deve-se usar ":" para separar chave e valor
# lembre-se de usar a virgula para separar items
# lembre-se que chaves podem ser qualquer tipo imutável de dados, no nosso caso são str mesmo

pessoa = {
    'nome' : 'wesley',
    'sobrenome': 'link',
    'idade' : 24,
    'altura' : 1.8,
    'endereços': [
        {'rua': 'itatiaia', 'numero': 187, 'casa': 4, 'cep': 79100390},
        {'rua': 'julio de castilhos', 'numero': 2841, 'casa': None, 'cep': 79100100},
    ],
}

# para verificar o valor de um determinado item, podemos acessar seu indice que é a própria chave
print(pessoa['nome'])
print(pessoa['sobrenome'])
print(pessoa['idade'])
print('-'*100)

# OBS: DICIONÁRIOS SÃO ITERÁVEIS, É POSSÍVEL USAR O 'FOR' NELES, veja:
for item in pessoa:
    print(item) # veja que dessa forma iteramos apenas sobre as chaves e não os valores

print('-'*100)

# para iterar sobre os valores podemos usar a seguinte técnica
for chave in pessoa:
    print(f'{chave=} {pessoa[chave]=}')
```

É  muito importante que saibamos manipular os valores contidos nos pares chave-valor de um dicionário python, para isso é necessário que saibamos iterar sobre os mesmos e manipulá-los através de laços e ou métodos/funções.

```python
pessoa = {}

chave = 'nome' # chave dinâmica, isso é: não é passada diretamente na declaração

pessoa[chave] = 'Luiz Otavio' # o valor da chave dinâminca é 'luiz otavio'
pessoa['sobrenome'] = 'miranda' # isso não é uma chave dinâmica

print(pessoa[chave])

pessoa[chave] = 'maria' #subscrevendo a chave

del pessoa['sobrenome'] # apagando chave
print(pessoa)
print(pessoa['nome'])

# forma de evitar uma excessão chamada KeyError que é disparada quando chamamos uma chave
# que não existe
if pessoa.get('sobrenome') is None: # método get tenta pegar o valor, se não achar return None
	print('não encontrei')
else:
	print(pessoa['sobrenome'])
```

COPIA RASA SHALLOW COPY E CÓPIA PROFUNDA DEEP COPY

    
Elementos mutáveis em python, como por exemplo listas e dicionários não podem ser copiados apenas usando uma atribuição normal igualando um iterável a outro de mesmo tipo, essa ação apenas diria ao interpretador que o novo objeto aponta pro endereço de memória do iterável mutável principal.

Para fazer cópias em python, nos mutáveis, precisamos fazer o que chamamos de shallow copy ou deep copy, que são cópias rasas e cópias profundas respectivamente… 

Porém, além do nome, elas possuem diferenças que são:

1. cópia rasa ou shallow copy: gera uma cópia da lista ou dicionário, porém de forma rasa, isto é, aplica a cópia de todos itens imutáveis da lista copiada, porém não copia os elementos mutáveis que podem estar dentro da lista copiada, veja:

```python
lista_1 = ['a','b','c','e']
lista_2 = lista_1[:] # podemos usar isso para gerar uma cópia raza

print(lista_1, lista_2)
lista_2.append('f')
print(lista_1, lista_2) # perceba que ocorre a mudança na lista 2 sem interferir na primeira

print()

lista_1 = ['a','b','c','e']
lista_2 = lista_1.copy() # podemos copiar de forma rasa desta maneira também.
lista_2.append('f')
print(lista_1, lista_2)


# AGORA PERCEBA O QUE A COPIA RASA FAZ
lista_1 = [1, 2, 3, ['a','b','c',]]
lista_2 = lista_1.copy() # fazendo uma cópia rasa
print(lista_1, lista_2) # veja que a cópia é feita

#podemos até alterar os dados mutáveis que serão modificados apenas em uma lista
lista_2[2] = 9
print(lista_1, lista_2)

#agora veja o que acontece quando alteramos os dados mutáveis mais internos em uma cópia rasa
lista_2[3].append('d') # farei um append apenas na sublista da lista_2 e isso será feito tb na lista_1
print(lista_1, lista_2)
```

2. cópia profunda ou deep copy em python é uma cópia que evita esse tipo de problema, ela é feita através do método deepcopy() do módulo copy embutido no próprio python, esse método retorna uma lista copiada de forma profunda, isto é, todos os itens são cópias, nenhum apontamento é feito, temos listas totalmente independentes. Veja:


```python
import copy # É NECESSÁRIO IMPORTAR O MÓDULO COPY

lista_1 = [1, 2, 3, ['a','b','c',]]
lista_2 = []

lista_2 = copy.deepcopy(lista_1) # fazendo cópia profunda
print(lista_1, lista_2)

lista_2[3].append('d')
print(lista_1, lista_2)
lista_1.remove(1)
print(lista_1, lista_2) # ambas listas continuam independentes
```

RESUMO DE CÓPIAS EM DICIONÁRIOS E LISTAS:

 Resumo:

    Shallow Copy: Copia apenas o objeto de nível superior e referencia os objetos aninhados. Use quando você deseja que as alterações nos objetos aninhados sejam refletidas em ambas as cópias.
    
    Deep Copy: Copia o objeto de nível superior e todos os objetos aninhados. Use quando você deseja que a cópia seja completamente independente do original.

EMPACOTAMENTO E DESEMPACOTAMENTO EM DICIONÁRIOS


O empacotamento e desempacotamento com dicionários é semelhante ao processo feito em listas e tuplas, porém agora precisamos lembrar que temos uma estrutura iterável do tipo chave-valor, portanto é necessário que lembremos sobre o que estamos iterando, se é sobre as keys, sobre os values, ou sobre os dois que seria iterar sobre os items.


```python
pessoa = {
    'nome': 'wesley',
    'sobrenome': 'Link',
    #'altura' : 1.6,
}

print(*pessoa) # da forma simples de desempacotamento o python desempacota somente as chaves

# para desempacotar desta maneira é necessário numero de variaveis igual ao numero de items

a, b = pessoa # irá mandar uma chave para cada valor
print(a)
print(b)

a, b = pessoa.values()
print(a)
print(b)

a, b = pessoa.items() # retorna uma tupla com os item
print(a)
print(b)

# desta forma eu consigo fazer um desempacotamento direto
(a1, a2), (b1, b2) = pessoa.items()
print(a1,a2)
print(b1,b2)
print()

# o trecho acima é semelhante a fazer isso:
for chave, valor in pessoa.items():
    print(chave, valor)

pessoa = {
    'nome': 'wesley',
    'sobrenome': 'Link',
    #'altura' : 1.6,
}

dados_adicionais = {
    'idade': 24,
    'altura':1.8,
}

# incluindo dicionário em outro
pessoa_completa = {**pessoa, **dados_adicionais, 'peso':80.00}
print(pessoa_completa)

# funciona em listas também
lista = [1,2,3,4]
lista_2 = [5,6,7,8]

lista_geral = [*lista, *lista_2]
print(lista_geral)
print()

#desempacotar um dicionário dentro de uma lista
print(pessoa_completa)
lista_geral = [*pessoa_completa.items()]
print(lista_geral)
```

# RESUMINDO *ARGS E **KWARGS - EMPACOTAMENTO DE ARGUMENTOS DE FUNÇÕES



Como dito no titulo, args e kwargs são usados para empacotamento de argumento em chamadas de funções, ou seja, são uma convenção usada na linguagem python que flexibiliza a chamada de função deixando o código mais limpo e dinâmico, possibilitando uma chamada de função simples que passa inúmeros argumentos posicionais ou então nomeados, veja:

```python
def funcao(*args, **kwargs):
    print('args: ', args, type(args))
    print('kwargs: ', kwargs, type(kwargs))

funcao(1,2,3,4, num5=5, num6= 6, num7=7) # PODEMOS PERCEBER QUE OS ELEMENTOS DO ARGS SÃO TUPLAS E DO KWARGS SÃO DICIONÁRIOS
print()
# imagine que poderiamos configurar uma tupla de argumentos nomeados para passar para a funcao
# podemos passar também argumentos nomeados usando um dicionario na chamada de funcao


args = (1,2,3,4) # lembre que são posicionais então a ordem importa
kwargs = {'num5':9, 'num6':10}
funcao (*args, **kwargs) # lembre-se: OS ARGUMENTOS POSICIONAIS RECEBEM ARGS (1 UNICO '*')
                         #            OS ARGUMENTOS NOMEADOS RECEBEM KWARGS (2 '*' )

# ambas chamadas de função são iguais, porém uma usa desempacomento e outra passa os argumentos
# de forma hard code (na unha)
```

PODEMOS USAR EMPACOTAMENTO EM CHAMADA DE FUNÇÕES MESMO QUE AS FUNÇÕES NÃO PREVENHAM O USO DE ARGS

Imagine uma função que obrigatoriamente requer 5 valores como argumentos posicionais que serão usados para uma formula matemática, esses valores podem ser muito grandes o que dificultaria a escrita da chamada de função, dessa forma podemos armazenar esses valores em listas ou tuplas, ou até mesmo sets e então passar esses valores na chamada de função usando desempacomento. veja:


```python
def formula_matematica(a, b, c, d, e, f):
    # embora a ordem na soma não importe, imagine que isso é uma formula que precisa receber
    # os valores na ordem passada
    soma = a+b+c+d+e+f
    print(soma)

# podemos chamar a função e passar os 6 argumentos na unha sempre, veja:
formula_matematica(125.5, 12.3, 45.5125, 1254265, 6.512564, 12)

# ou podemos elegantemente gerar um iterável disso
lista_valores = [
    125.5, 
    12.3,
    45.5125, 
    1254265, 
    6.512564, 
    12
]

tupla_valores = (
    125.5, 
    12.3,
    45.5125, 
    1254265, 
    6.512564, 
    12.3
)

set_valores = {
    125.5, 
    12.3,
    45.5125, 
    1254265, 
    6.512564, 
    12
}

# desempacotamento usando listas
print('usando listas')
formula_matematica(*lista_valores)
print()

# desempacotamento usa tuplas
print('usando tuplas')
formula_matematica(*tupla_valores)
print()

# desempacotamento usa sets
print('usando sets')
formula_matematica(*set_valores)
print()
```
