# FUNÇÕES LAMBDA EM PYTHON

São funções de uma linha que não recebem nome, por estes motivos são chamadas de funções anônimas e são muito utilizadas em operações rápidas e simples.

Veja abaixo exemplos de uso:


```python
# FUNÇÕES LAMBDA

# sintaxe básica
lambda argumentos: expressao

# exemplo simples:
soma = lambda x, y: x+y
print(soma(2,3))
```

O retorno de uma função lambda em Python é a expressão que você define após os dois pontos (:). A expressão é avaliada e seu resultado é retornado automaticamente. Não é necessário usar a palavra-chave return como em funções normais.

```python
# a funcao lambda não tem nome, o que vem antes dos ':' é o argumento
# o que vem depois é o próprio retorno
quadrado = lambda x: x**2
print(quadrado(2))

# exemplo de transcrição de uma função definida para uma função lambda
# reparem que o resultado será o mesmo em ambas

# funcao definida
def maior_definida(x,y):
    return x if x > y else y # operação ternária de retorno

# funcao lamda com operação ternária 
maior_lambda = lambda x, y: x if x > 2 else y


print(maior_definida(2,3))
print(maior_lambda(2,3))
```

FUNÇÕES LAMBDA SÃO USADAS EM FUNÇÕES DE ALTA ORDEM

As funções lambda são frequentemente usadas em funções de alta ordem para definir a regra de negócio da função de alta ordem, como por exemplo as funções map(), filter() e reduce() que são funções embutidas responsáveis por algumas ações.

A função lambda pode ser usada com a função sorted() no argumento key para definir a chave de ordenação de uma lista.

```python
# ordenando conjuntos simples com o método sort()
numeros = [1, 2, 5, 4, 3]
numeros.sort() # método que ordena uma lista
print(numeros)

pessoas = ['laura', 'amaral', 'link', 'wesley', 'zidane']
pessoas.sort()
print(pessoas) # veja que ele ordena por ordem alfabética
```

```python
# em conjunto de dados simples o método sort() funciona ok, porém imagine
# que tenhamos estruturas complexas, com dados misturados, inteiros, floats, str
# ou então que tenhamos um dicionário com pares chave-valor, como o python deve
# ordenar? Ai que entra as funções lambda que vão passar a regra de negócio para
# a função de alta ordem

# o python não sabe como ordenar um dicionário
# eu quero ordem em ordem alfabetica pelo nome
# ou eu quero ordenar pelo sobrenome


lista = [
    {'nome': 'Luiz', 'sobrenome': 'miranda'},
    {'nome': 'Maria', 'sobrenome': 'Oliveira'},
    {'nome': 'Daniel', 'sobrenome': 'Silva'},
    {'nome': 'Eduardo', 'sobrenome': 'Moreira'},
    {'nome': 'Aline', 'sobrenome': 'Souza'},
    {'nome': 'Amanda', 'sobrenome': 'miranda'},
]

# definindo a função no proprio parametro
ordenado_por_nome = sorted(lista, key=lambda x: x['nome'])
print(*ordenado_por_nome, sep='\n')

print()

# definindo a função por fora e passando o função no parâmetro
funcao_ordena_nome = lambda lista_pessoas: lista_pessoas['nome']
ordenado_por_nome = sorted(lista, key=funcao_ordena_nome)
print(*ordenado_por_nome, sep='\n')
```

    FUNÇÃO MAP ()

    A função map() aplica uma função a todos itens de uma lista, dicionário, tupla, ou qualquer iterável e retorna um novo iterável com os resultados.

    map(funcao, iterável)

```python
# usando map(), map aplicará uma função (geralmente lambda, a cada item do iterável)
# usar um iterável dentro de outro iterável e aplicar map nesses casos, pode ser complexo

numeros = [1,2,3,4]
quadrados =  list(map(lambda x: x**2, numeros))
print(quadrados)
print()

lista_tuplas = [
    (1,2,3,4,5),
    (2,3,4),
    (1,2)
]

lista_tuplas_metades =  list()
for tupla in lista_tuplas:
    metades = tuple(map(lambda x: x/2, tupla))
    lista_tuplas_metades.append(metades)

print(*lista_tuplas_metades, sep='\n')
print()

# mesmo laço sem aplicar map e funções lambda:
lista_tuplas_metades = list()
for tupla in lista_tuplas:
    lista_dividida = list()
    for indice in tupla:
        metade_indice = indice / 2
        lista_dividida.append(metade_indice)
    tupla_dividida = tuple(lista_dividida)
    lista_tuplas_metades.append(tupla_dividida)

print(lista_tuplas_metades)
    
        
```

FUNÇÃO FILTER ()

A função filter() aplica um filtro ao iterável e retorna outro iterável contendo apenas os itens os quais foram avaliados como True pela funcao lambda passada no argumento.

filter (funcao, iterável)

```python
lista = list(range(0, 100))

lista_pares = list(filter(lambda x: (x % 2) == 0, lista))
set_impares = set(filter(lambda x: (x % 2) != 0 , lista)) # usando sets somente para mostrar que posso usar qualquer iterável
print(lista_pares)
print(set_impares)
```

FUNÇÃO REDUCE()

Função cujo objetivo é reduzir os itens de uma sequencia à um único valor conforme função passada como argumento, diferente da função map() que aplica uma função a cada item de uma sequencia e retorna um iterável com as alterações, a função reduce() aplica de forma cumulativa uma regra à uma sequência e retorna um valor.

sintaxe básica:

reduce(funcao, sequencia, [valor_inicial])

funcao: Uma função que recebe dois argumentos e retorna um único valor.
sequencia: A sequência de itens a serem processados.
valor_inicial (opcional): Um valor inicial para o acumulador.

```python
from functools import reduce

numeros = [10,1,2,3,2,4,5,9]

def soma(x,y):
    return x+y

# aplicando o reduce
resultado = reduce(soma, numeros)
print(resultado)

def multiplicar(x,y):
    return x*y

resultado = reduce(multiplicar, numeros)
print(resultado)

# podemos usar valores iniciais também
resultado = reduce(multiplicar, numeros, 2) # valor inicial é 2
print(resultado) # [2,1,2,3,4,5]

# achar o maior numero de uma sequencia
maior_numero = reduce(lambda x,y: x if x > y else y, numeros)
print(maior_numero)

# mostrando como é possível fazer mais simplificadamente
maior_numero = max(numeros)
print(maior_numero)
```
