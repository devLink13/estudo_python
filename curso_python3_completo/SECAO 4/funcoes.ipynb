# FUNÇÕES 'DEF' EM PYTHON

Funções no python, são criadas e definidas como ‘def’, elas servem basicamente para o reuso de um código, como assim? quando queremos exibir algo na tela, usamos a função embutida print, ela nada mais é que uma função def que foi criada anteriormente por alguém e serve para evitar a manipulação de muitos trechos de códigos e que apenas retorna um valor que será exibido na tela.

Resumindo, funções são trechos de código reutilizáveis e que servem para realizar alguma ação. No python as funções por padrão retornam None, isto é, um ‘não valor’, ou seja, retornam nada; porém isso não impede que ao criar a função retornemos um valor com um tipo diferente de none, podendo ser bool, string, int ou float ou qualquer outro dado.

Funções podem receber parâmetros que serão usados dentro do escopo da função para realizar alguma ação, esses parâmetros são variáveis basicamente, quando recebem um valor, são chamados de argumento, logo um parâmetro recebe um argumento. Os argumentos podem ser posicionais, isto é, são obrigatórios, ou podem ser opcionais, o que muda é a forma de declaração do parâmetro.

```python
#  função imprimir que necessita de 3 argumentos POSICIONAIS (NÃO NOMEADOS)
def imprimir(a , b, c):
	print(a, b, c)
	
imprimir(1, 2, 3) # argumentos = 1, 2 e 3 respectivamente

# função com argmento opcional- isso é um argumento NOMEADO
def saudacao(nome='Sem nome'):
	print(f'OLÁ, {nome}.')
	
saudacao('LINK')
saudacao()
```

```python
# A ORDEM DOS ARGUMENTOS POSICIONAIS - NÃO NOMEADOS - FAZ DIFERENÇA

def somar(x, y):
    print(f'{x=} {y=}  | {x} + {y} = {x+y}')

#     x  y
somar(1, 2)
somar(2, 1)

print()

#se eu passar o argumento posicional nomeando ele ai tanto faz a ordem
somar(y=2, x=1)
somar(x=1, y=2)
```

```python
# após usar um argumento nomeado, todos os argumentos subsequentes deverão ser nomeados

def somar(x, y):
    print(f'{x=} {y=}  | {x} + {y} = {x+y}')

#     x  y
somar (1, y=2) # isso funciona ok
somar (y=2, 1) # isso gera um erro
```

CUIDADO COM VALORES PADRÃO

É necessário ter atenção á definicação de valores padrão para os parâmetros, isso pode gerar problemas na lógica do código
veja abaixo:

```python
# isso funciona ok, porém e se quisermos somar o z somente quando ele for enviado?
def somar(x, y, z = 0):
    print(f'{x=} {y=} {z=}  | {x} + {y} + {z} = {x+y+z}')

#     x  y
somar (1, y=2) # isso funciona ok
print()

#cuidado ao comparar o z no if, pois z recebe 0 por padrão, e zero é False
# a função abaixo possui bugs
def somar(x, y, z = 0):

    if z: # se z for enviado
        print(f'{x=} {y=} {z=}  | {x} + {y} + {z} = {x+y+z}')
        print('tem z')
    
    else:
        print(f'{x=} {y=} | {x} + {y} = {x+y}')
        print('nao tem z')

somar(1, 2) # não enviei o z
print()
somar(1,2,3) # enviei o z e funcionou ok
print()

#agora perceba o bug, eu vou enviar o z como 0, e ele não executará o código como se eu tivesse mandado, veja:
somar(1,2,0) # me informa que não tem z, mas claramente tem.
print()

# FORMA DE RESOLVER ISSO É USAR O NONE -> não valor
def somar(x, y, z = None):

    if z is None: # se z for enviado
        print(f'{x=} {y=} | {x} + {y} = {x+y}')
        print('nao tem z')
    
    else:
        print(f'{x=} {y=} {z=}  | {x} + {y} + {z} = {x+y+z}')
        print('tem z')
        
somar(1,2,0) #agora ele considera o z e informa que tem z
print()

somar(1, 2) #funciona corretamente sem considerar o z
print()
```

ESCOPO DE FUNÇÕES


O escopo é o lugar onde o código é alcançável, isto é, onde ele é acessível. De modo mais simplista, podemos dizer que o escopo é o “mundinho” fechado de um treco específico de código. Quando falamos em escopo estamos falando sobre o que temos alcance em cada local, lembrando que cada função, cada modulo e cada arquivo tem seu próprio escopo, podendo ele ser global (comum a todo código), pode ser interno a função x, interno a função y ou até mesmo interno à algum laço de repetição. Veja abaixo:

```python
# aqui temos o escopo global
'''
    TUDO QUE ESTIVER NESTE ESCOPO PODE SER ACESSADO POR QUALQUER FUNÇÃO	
    E QUALQUER LUGAR DO PROGRAMA
'''

x = 10
y = 20

def printar():
    #aqui temos o escopo local da função somar
    '''
         TUDO QUE ESTIVER DENTRO DESSA FUNÇÃO SÓ PODE SER ACESSADO DENTRO DELA
         OU POR FUNÇÕES DENTRO DELA
    '''

    # aqui temos as variaveis x e y sendo declaradas com outro valor
    # porém notemos que o nome é o mesmo do escopo global, isto não é um problema
    # pois o python entende que são variáveis diferentes e não sobrescreve a variável global
    # porém se eu quiser acessar a variável global dentro da função, eu devo usar a palavra reservada global

    x = 1
    y = 2

    # irá printar o x e y local
    print(x, 'escopo local') 
    print(y, 'escopo local')

printar()
print(f'{x=} escopo global | {y=} escopo global')
```

```python
# também podemos entender o uso da palavra reservada global

x = 10
y = 20

def printar():
    #aqui temos o escopo local da função somar
    '''
         TUDO QUE ESTIVER DENTRO DESSA FUNÇÃO SÓ PODE SER ACESSADO DENTRO DELA
         OU POR FUNÇÕES DENTRO DELA
    '''

    # aqui temos as variaveis x e y sendo declaradas com outro valor
    # porém notemos que o nome é o mesmo do escopo global, isto não é um problema
    # pois o python entende que são variáveis diferentes e não sobrescreve a variável global
    # porém se eu quiser acessar a variável global dentro da função, eu devo usar a palavra reservada global

    global x
    global y

    # irá printar o x e y local
    print(x, 'escopo local') 
    print(y, 'escopo local')

printar()
print(f'{x=} escopo global | {y=} escopo global')
```

# ATENCAO AO USO DE LAÇOS FOR E WHILE EM PYTHON EM RELAÇÃO AOS SEUS ESCOPOS

    em Python, o laço for não cria um novo escopo. As variáveis definidas dentro de um laço for são acessíveis fora do laço, desde que estejam no mesmo escopo da função ou do bloco em que o laço está inserido.

    o laço while também não cria um novo escopo. As variáveis definidas dentro de um laço while são acessíveis fora do laço, desde que estejam no mesmo escopo da função ou do bloco em que o laço está inserido.

```python
for i in range(10):
    contador = i
    print(i)

print(contador) # funciona porque em python o for não cria um escopo local
```

```python
contador = 0
while contador < 10:
    print(contador)
    contador += 1

    palavra = ''+str(contador)

print(palavra) # funciona porque o while não cria escopo local
```

OUTROS ESTRUTURAS QUE NÃO CRIAM NOVO ESCOPO EM PYTHON SÃO:

    1. laços for e while
    2. blocos condicionais if, elif e else
    3. blocos de excecção try except

AFINAL QUANDO CRIAMOS UM NOVO ESCOPO EM PYTHON?

    Em resumo, em Python, apenas funções, métodos e classes criam novos escopos locais. Estruturas de controle de fluxo como for, while, if, try, compreensões de listas, dicionários e conjuntos não criam novos escopos, e as variáveis definidas dentro dessas estruturas são acessíveis fora delas, desde que estejam no mesmo escopo da função ou do bloco em que foram definidas.


```python
# vamos verificar algumas coisas sobre os escopos em estruturas do python

for i in range(10):
    contador = i

print(contador) # funciona porque em python o for não cria um escopo local

if 10 > 5:
    palavra = 'maior'
print(palavra) # funciona porque o if não cria um escopo local

try:
    x = 10
except:
    y = 20
finally:
    z = 30

print(x, y, z) # funciona porque o try, except e finally não criam escopo local

```

```python
# NO PRIMEIRO TRECHO DE CÓDIGO ABAIXO VEREMOS QUE É POSSÍVEL ACESSAR A VARIÁVEL DO ESCOPO DE UM WHILE
# QUANDO FORA DO ESCOPO DE UMA FUNÇÃO, MAS NO SEGUNDO TRECHO DE CÓDIGO VEREMOS QUE NÃO É POSSÍVEL ACESSAR
# A VARIÁVEL DO ESCOPO DE UM WHILE QUANDO ESTE ESTIVER DENTRO DE UMA FUNÇÃO

# vamos ver o código fora de uma função
contador = 0
while contador < 10:
    print(contador)
    contador += 1

    palavra = ''+str(contador)

print(palavra) # funciona pois while não cria escopo local e não está dentro de uma função

# agora vamos ver o mesmo código, porém dentro de uma função

def printar():
    contador = 0
    while contador < 10:
        print(contador)
        contador += 1

        palavra_2 = ''+str(contador)

print(palavra_2) # "palavra 2 não está definida" pois o while cria escopo local e está dentro de uma função
```

#  MÉTODOS E FUNÇÕES PARA VERIFICAR ESCOPOS

Existem algumas funções que podem retornam informações para nós que serão úteis na obtenção do escopo de um determinado dado. veja abaixo:

A FUNÇÃO DIR() É A MAIS INTERESSANTE DE TODAS PARA VERIFICAR UMA LISTA NO ESCOPO ATUAL OU ENTÃO UMA LISTA DE ATRIBUTOS DE UM DETERMINADO OBJETO INSTANCIADO.

```python
# função locals() retorna um dicionário com todas as variáveis locais

x = 10
y = 20

def printar():
    w = 1
    z = 2
    print(locals()) # imprime um dicionário com todas as variáveis locais que seriam w e z

printar()
print(locals()) # imprime um dicionário com todas as variáveis locais que seriam x e y

```

```python
# função globals() retorna um dicionário com todas as variáveis globais


x = 10
y = 20

def printar():
    w = 1
    z = 2
    print(globals()) # imprime um dicionário com todas as variáveis globais que seriam x e y

printar()
print(globals()) # imprime um dicionário com todas as variáveis globais que seriam x e y
```

```python
# a função vars()  retorna o __dict__ de um objeto, desta forma podemos verificar as variváveis de um objeto.
# exemplo de uso da função vars (), usada para sada para retornar o dicionário de atributos de um objeto. 
# Se nenhum argumento for passado, ela retorna o dicionário de variáveis locais no escopo atual.

x = 10
y = 20

soma = x + y
print(soma)
print(vars()) # retorna as variaveis do escopo local
print('-'*50)

class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade
        self.sexo = 'masculino'

# criando um objeto da classe pessoa chamado wesley
wesley = Pessoa('wesley', 24)
atributos_wesley = vars(wesley)

print(atributos_wesley)
```

```python
# a função dir() é usada para tentar retornar uma lista de atributos válidos para um objeto. 
# Se nenhum argumento for passado, ela retorna a lista de nomes no escopo local atual.
# Aqui está um exemplo de como usá-la:

x = 10
y = 20

soma = x + y
print(soma)

# Usando dir() para listar os nomes no escopo local
nomes_locais = dir()
print(nomes_locais) # retorna os nomes no escopo local
print('-'*50)

for item in dir():
    print(item)
```

# RETORNO DE FUNÇÕES - RETURN-
Toda função retorna retorna algo, no python por padrão retorna None, que é um não valor, toda função se não for definido um return ela irá retornar None, independente das ações feitas.


```python
def saudacao(nome): # essa função printa uma saudação mas retorna None
	print(f'olá {nome}')
	
saudacao('wesley') # embora essa função execute um print, ou uma ação, ela retorna none
valor = saudacao('wesley') 
print(valor) # None

def somar(x, y)-> float: # na definição da função após a seta  podemos perceber o que ela retorna, que é um float
	soma = x + y
	return soma # o retorno será a soma de x e y

soma = somar(1.5 , 2)
print(soma, type(soma))
```

*ARGS - CONVENÇÃO PARA EMPACOTAMENTO EM DECLARAÇÃO DE FUNÇÕES.

*args é uma convenção no mundo do python para usar empacotamento de argumentos não nomeados na declaração de uma função que necessite de múltiplos argumentos. Veja:

```python
# estou usando o *args para empacotar todos os dados em uma tupla
def somar(*args):
    total = 0
    for numero in args:
        total = total + numero
    return total


soma = somar(1,2,3,4)

#repare que se eu passar os valores como um iterável(por exemplo uma tupla)
#a função recebe uma tupla dentro de outra tupla
numeros_para_somar = (1,2,3,4)
#somar(numeros_para_somar) # aqui eu gero um typeError 

# para resolver isso usamos o DESEMPACOTAMENTO
soma1 = somar(*numeros_para_somar) # usando o desempacotamento
soma2 = somar(1,2,10)

print(soma, soma1, soma2, sep=' / ')
```

CLOSURE E FUNÇÕES QUE RETORNAM FUNÇÕES


```python
# A IDEIA AQUI É ENTENDER O CONCEITO DE CLOSURE EM PYTHON E O USO DE FUNÇÕES QUE RETORNAM FUNÇÕES

'''
    CONCEITOS:
        A) CLOSURE: É O FECHAMENTO DE UMA FUNÇÃO EM TEMPO DE CHAMADA.
        B) FUNÇÕES QUE RETORNAM FUNÇÕES: SÃO FUNÇÕES QUE RETORNAM OUTRAS FUNÇÕES
'''

# VAMOS CRIAR UMA FUNÇÃO QUE RETORNA UMA FUNÇÃO
def criar_saudacao(saudacao):
    def saudar(nome):
        return f'{saudacao} {nome}'
    return saudar # veja que estou retornando apenas a função saudar, sem o uso de parenteses, pois se eu usar os parenteses estarei
                  # resolvendo a funcao e retornando o resultado dela na hora, e não é isso que queremos

# repare que abaixo eu vou criar duas funções que recebem um argumento e geram outra função a partir daí, o conteudo dessa função é armazenado
# nas variáveis falar_bom_dia e falar_boa_noite respectivamente, e ainda naõ foram resolvidas, ou seja, ainda não foram chamadas
# portanto eu posso chamar essas funções depois usando os ()

# repare também que faltará um argumento, o argumento que falta é o nome, que será passado quando eu chamar a função retornada

falar_bom_dia = criar_saudacao('bom dia') # aqui eu armazeno o conteúdo retornado pela função criar_saudacao sem resolver a funcao ainda
falar_boa_noite = criar_saudacao('boa noite') # aqui eu armazeno o conteúdo retornado pela função criar_saudacao sem resolver a funcao ainda

# o closure acontece quando eu chamo a função falar_bom_dia ou falar_boa_noite, e passo o argumento nome usando os parenteses
# o closure é o fechamento da execucação da função que é feito usando os '()' após o nome das variáveis

print(falar_bom_dia(nome='Laura')) # aqui eu chamo a função criar_saudacao e passo o argumento nome
print(falar_boa_noite(nome='Laura')) # aqui eu chamo a função criar_saudacao e passo o argumento nome


'''
    RESUMO:
        Perceba que apartir de uma função eu criei duas funções que retornam uma saudação, uma de bom dia e outra de boa noite
        o uso do closure, da função que retorna função e da reutilização de função permite que criemos muitas funções a partir de uma só
'''
```
