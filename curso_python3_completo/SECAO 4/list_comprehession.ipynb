# LIST COMPREHENSION EM PYTHON

É um artifício do python usado para gerar novas listas a partir de iteráveis usando uma sintaxe específica e aplicando conceitos de mapeamento e filtro numa única linha de comando.

Resumindo: List comprehension em Python é uma maneira concisa e eficiente de criar listas. Ela permite construir listas de uma forma mais compacta e legível, utilizando uma única linha de código. A sintaxe básica é:

[inclua_isso for item in iterável if condicao] → lembrando que tudo que está a esquerda do for é mapeamento e tudo que está a direita é filtro.

```python
# criando uma lista de numeros de 0 a 9 usando list comprehension
numeros = [x for x in range(10)]

```

List comprehensions são úteis para simplificar o código e torná-lo mais legível, especialmente quando comparado com a criação de listas usando loops `for` tradicionais. Porém as coisas podem começar a ficar complexas se aplicarmos muitas coisas no meio do caminho.

````python
lista = list(range(50))
# print(lista)

# o que vem a esquerda do for é a expressão que inclue na nova lista os valores
# isso é chamado de mapeamento

lista_dobro = [numero * 2 for numero in lista]
print(lista_dobro) # geramos uma outra lista a partir da primeira, muito parecido
                   # com a função map() que aplica uma função a cada item do iteravel

# aplicar o dobro apenas nos pares, repare que estamos fazendo um mapeamento e um filtro
lista_dobro_PARES = [numero * 2 for numero in lista if (numero % 2) == 0]
print(lista_dobro_PARES)

'''
    # OBS: NO LADO DO MAPEAEMENTO (A ESQUERDA DO FOR, SEMPRE USAMOS IF E ELSE (OPERAÇÃO TERNÁRIA)
           JÁ NO LADO DO FILTRO, À DIREITA DO FOR, SOMENTE O IF JÁ BASTA
'''

# passando uma função lambda para a list comprehension
lista_dobro_impares = [(lambda numero: numero*2) (numero) for numero in lista if (numero % 2) != 0]
print(lista_dobro_impares)


# passanndo uma função definida na list comprehension
def quadrado(numero):
    return numero ** 2

lista_quadrados = [quadrado(numero) for numero in lista]
print(lista_quadrados)
````

````python
# EXEMPLO> USAR LIST COMPREHENSION PARA CRIAR UMA LISTA A PARTIR DE OUTRA
#          MODIFICANDO ALGUNS PARÂMETROS SE DETERMINADA LOGICA FOR SATISFEITA
lista_produtos = [
    {
        'nome': 'iphone 15',
        'preço': 5433.50,
        'categoria': 'eletrônicos',
    },
    {
        'nome': 'batedeira',
        'preço': 500.00,
        'categoria': 'eletrodomestico',
    },
    {
        'nome': 'macbook',
        'preço': 12000,
        'categoria': 'eletrônicos',
    },
    {
        'nome': 'xbox series s',
        'preço': 2500.00,
        'categoria': 'eletrônicos',
    },
]

# print(*lista_produtos, sep='\n')

# crie uma nova lista de produtos promocionais dando desconto de 50% em itens da categoria eletrônicos usando uma unica linha de código
lista_promocao_eletronicos = [{**item, 'preço': item['preço']*0.5} for item in lista_produtos if item['categoria'] == 'eletrônicos']
print(lista_promocao_eletronicos)
print()

# o mesmo código acima usando map()
lista_promocao_eletronicos = list(map(lambda item: {**item, 'preço': item['preço'] * 0.5} if item['categoria'] == 'eletrônicos' else None, lista_produtos))
print(lista_promocao_eletronicos) # repare que a lista contará com None onde não houve modificações, para evitar isso seria melhor usar a listcomprehension
print()

# REPARE QUE NÃO ADIANTA USAR FILTER TAMBÉM, LOGO, A LIST COMPREHESION É A MELHOR COISA DO MUNDO NESTES CASOS QUE PRECISAMOS FILTRAR E MODIFICAR LISTAS
lista_promocao_eletronicos = list(filter(lambda item: {**item, 'preço':item['preço']*0.5} if item['categoria'] == 'eletrônicos' else False, lista_produtos))
print(lista_promocao_eletronicos)
````
