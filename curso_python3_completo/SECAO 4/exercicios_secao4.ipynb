1. CRIAR UMA FUNÇÃO QUE MULTIPLIQUE TODOS ARGUMENTOS NÃO NOMEADOS RECEBIDOS E RETORNAR O TOTAL PARA UMA VARIAVEL
MOSTRANDO NA TELA O RESULTADO


```python
def multiplicacao(*args):
    total = 1
    for numero in args:
        total = total * numero
    return total

numeros = 2,3,5,2
resultado = multiplicacao(*numeros)
print(resultado)
```

2. CRIAR UMA FUNÇÃO QUE VERIFICA SE O NUMERO É PAR OU IMPAR, RETORNAR ESTE VALOR E MOSTRAR NA TELA.

```python
#caso queiramos retornar um literal contendo uma string
def impar_par(numero):
    resultado = 'par' if (numero % 2) == 0 else 'impar'
    return resultado

impar_ou_par = impar_par(2)
print(impar_ou_par)

impar_ou_par = impar_par(593)
print(impar_ou_par)
```

```python
# caso queiramos retornar um bool
def impar_par(numero):
    return (numero % 2) == 0

print(impar_par(2))
print(impar_par(3))
```

3. CRIAR UMA FUNÇÃO COM USO DE CLOSURE, FUNCOES QUE RETORNAM FUNÇÕES PARA DUPLICAR, TRIPLICAR E QUADRUPILHAR O NUMERO PASSADO COMO PARÂMETRO PARA UMA FUNÇÃO 'PAI'.

```python
# criar uma função genérica
def multiplicar_por(multiplicador): # essa função recebe por quantas vezes queremos multiplicar algum numero

    # criar uma função que retorne a operação que o usuário quiser
    # permitindo o reuso de código
    def multiplicar(numero): # esta função recebe o numero que queremos multiplicar pelo multiplicador
        return multiplicador * numero
    
    return multiplicar


duplicar = multiplicar_por(2)
triplicar = multiplicar_por(3)
quadruplicar = multiplicar_por(4)
print(type(duplicar), 'veja que duplicar é uma função') #ela foi gerada a partir de uma função genérica



#fazendo closure
print('duplicando o numero 3 = ', duplicar(3) )
print('triplicando o numero 3 = ', triplicar(3) )
print('quadruplicar o numero 3 = ', quadruplicar(3) )



```
