# OPERAÇÃO TERNÁRIA
Operações ternárias são basicamente if e else em uma única linha seguindo alguns padrões de sintaxe, sendo eles da seguinte forma: este_valor_atribuido se  condicao_verdadeira senao for atribuir_este_valor. No código abaixo fica mais explícito.

```python
# IMAGINEMOS UMA CONDIÇÃO SIMPLES QUE POSSA SER EXPRESSA RAPIDAMENTE EM UM VALOR OU OUTRO

'''
Na verdade é possível aninhar condições ternárias, porém isso não é recomendado !!!
'''

condicao = True
#condicao = False

#leiamos: atribuir 10 se condição for verdadeira, senão atribuir 'condicao insatisfeita'
resultado = 10 if condicao == True else 'condição insatisfeita' 
print(resultado)

# NÃO FAZER ISSO QUE SERÁ MOSTRADO ABAIXO
nota1 = 3
nota2 = 4
media = ((nota1 + nota2)/2)

situacao = 'aprovado' if media >= 6 else 'recuperacao' if media >= 4 and media < 6 else 'reprovado'
print(situacao, f'{media=}')
```

```python
# o ideal é atribuir valores em operações rápidas, lembre-se que simples é melhor que complexo

while True:
    idade = input('digite sua idade: ')
    try:
        idade_int = int(idade)
    except:
        print('digite uma idade válida usando apenas números...')
        continue

    situacao = 'maior' if idade_int >= 18 else 'menor de idade'
    situacao_invertida = 'menor de idade' if idade_int < 18 else 'maior'

    if situacao:
        break

print(f'{situacao=}', f'{situacao_invertida=}')
```
