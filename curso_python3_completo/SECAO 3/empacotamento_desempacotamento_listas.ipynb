### DESEMPACOTAMENTO DE LISTAS

Uma prática muito comum em python é o desempacotamento de um pacote em alguma variável, lembrando que tipos que contém vários dados são pacotes, como listas, tuplas e dicionários…

Desempacotar é o ato de coletar determinado valor de um índice específico do pacote e fazer algo com o resto, veja:

```python
letras = ['a', 'b', 'c']
letra1, letra2, letra3 = letras
print(letra1, letra2, letra3) #'a' 'b' 'c'

letra1, letra2, letra3 = ['a', 'b', 'c']
print(letra1, letra2, letra3) #'a' 'b' 'c'
```

repare que o numero de variáveis corresponde ao número de valores, agora imagine que exista um lista muito extensa de valores e que queiramos coletar apenas o 2 valor, como faríamos?

```python
letras = ['a', 'b', 'c', 'd', 'e']

_, letra2, *resto = letras
print(f'{_=}, {letra2=}, {resto=}')
```

O operador de empacotamento é o ‘*’ que serve para acumular o restante dos valores numa outra variável que agora se chama resto, caso não tivéssemos feito isso o python estouraria um erro de ValueError, portanto, sempre que usarmos desempacotamento é necessário dizer ao python o que deverá ser feito com o restante do pacote, tanto antes, quanto depois do valor desempacotado, repare que antes da letra2, foi usado ‘_’ que é uma convenção em python para indicar uma variável não usada, ela foi responsável por receber o valor da primeira letra.

```python
letras = ['a', 'b', 'c', 'd', 'e']
_,_, letra3, *resto = letras

print(letra3, resto)
```


COMO DESEMPACOTADOR VARIOS VALORES INICIAIS, PEGAR UM VALOR MEDIANO E NOVAMENTE EMPACOTAR O RESTO?

para isso é necessário usar o fatiamento, não esquecendo do operador de empacotamento em nenhum momento, se não usarmos o operador
de empacotamento no momento do fatiamento, o resto será uma lista de lista, se isso não for problema, podemos ignorar este fato, veja:


```python
letras = ['a', 'b', 'c', 'd', 'e']
letras1_2, letra3, *resto = letras[:2], letras[2], *letras[3:]
print(f'{letras1_2=} {letra3=} {resto=}')

# sem usar o '*' no fatiamento, veja:
letras = ['a', 'b', 'c', 'd', 'e']
letras1_2, letra3, *resto = letras[:2], letras[2], letras[3:]
print(f'{letras1_2=} {letra3=} {resto=}')
```

geralmente o resto de uma lista é descartado, se esse for o caso podemos desconsiderar a atenção no operador de empacotamento no momento do fatiamento

DESEMPACOTAMENTO EM CHAMADA DE FUNÇÕES

É basicamente fazer o desempacotamento de um pacote ou iterável na própria chamada de função usando o operador de desempacotamento, sem a necessidade de usar um laço de repetição, veja:

```python
lista = ['laura', 'wesley', [0,1,2,3], 'josé', 'maria', 'lucas']

# o normal seria da seguinte forma:
for nome in lista:
    print(nome)

# desempacotar na chamada de função:
print(*lista, sep=' - ')
print()

print('desempacotamento na chamada de função: ')
print(*lista, sep='\n') # usando o sep como uma quebra de linha, fica interessante visualizar os índices
```

```python
# um pouco mais de empacotamento e desempacotamento
lista = [1,2,3,4]
lista_2 = [5,6,7,8]

lista_geral = [*lista, *lista_2]
print(lista_geral)
```
