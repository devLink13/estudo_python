## A IMPRECISÃO DOS NÚMEROS DE PONTO FLUTUANTE

Algumas linguagens de programação, como o python, possuem dificuldades para entregar uma precisão nos resultados de operações com ponto flutuante, isso ocorre devido a problemas de referenciamento de memória. Porém isso não é tão relevante uma vez que conseguimos contornar esses problemas usando alguns artifícios.

```python
numero1 = 0.1
numero2 = 0.7
soma = numero1 + numero2 # resultado esperado é 0.8
# veja a imprecisão
print(soma) #0.7999999999999999
```

```python
# formas de contornar isso
# a) usando f-strings na formatação
numero1 = 0.1
numero2 = 0.7
soma = numero1 + numero2 # resultado esperado é 0.8

# aplicando as f-strings conseguimos formatar o número e entregá-lo formatado
print(f'{soma=:.2f}', type(f'{soma=:.2f}'))
```

Qual o problema de contornar isso dessa maneira?
    
    é que o tipo do dado agora torna-se uma string, pois as f-strings retornam uma string formatada, dito isso, podemos prever que a soma continue sendo um numero impreciso, para contornar o problema trabalhando propriamente no número nós precisamos usar métodos e funções de arredondamento.

```python
# usando função round()
numero1 = 0.1
numero2 = 0.7
soma = numero1 + numero2 # resultado esperado é 0.8

soma = round(soma, 2)
print(soma)
```

usar o módulo decimal com a função Decimal: O módulo nativo decimal que vem junto com o python também pode nos ajudar a resolver esse problema, porém é um pouco mais complexo de se utilizar e também requer um pouco mais de atenção, uma vez que para funcionamento correto e assertivo da função nós precisamos passar strings ao invés de números de ponto flutuante, veja:

```python
import decimal
numero1 = decimal.Decimal('0.1')
numero2 = decimal.Decimal('0.7')

soma = numero1 + numero2
print(soma, type(soma)) #0.8 #atençao ao tipo de dado retornado
soma = float(soma)
print(soma, type(soma)) # é necessária a coerção de tipo após a utilização da função
```
