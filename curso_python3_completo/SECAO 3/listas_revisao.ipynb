# ABAIXO VEJA O USO DE MÉTODOS ÚTEIS

      append - Adiciona um item ao final
      insert - Adiciona um item no índice escolhido
      pop - Remove do final ou do índice escolhido
      del - apaga um índice
      clear - limpa a lista
      extend - estende a lista
      + - concatena listas

```python
# uso do append
lista = [1, 12.3, 'abc']
print(lista)
lista.append('d')
print(lista) # [1, 12.3, 'abc', 'd'] 
```

    # uso do método del: cuidado ao utilizar em grandes listas, podem consumir muito recurso computacional

O método del retira um elemento específico da lista, e depois reorganiza a lista, pensando desta maneira, se tivermos uma lista 
com milhares de elementos, o python deverá deslocar milhares de elementos após remover um ítem do meio desta lista.

```python
# método del
lista = [1, 12.3, 'abc']
print(lista)
del lista[1] #removendo o elemento do indice 1
print(lista)
print(lista[1]) #veja que ainda há elemento no índice 1, o python reorganizou a lista
```

Veja abaixo que o método pop remove normalmento no final da lista, porém se passarmos como parâmetro o índice de um elemento, podemos remove-lo e guardar o seu valor se for necessário.

```python
# método pop
lista = [1, 12.3, 'abc']
print(lista)
lista.append('d')
print(lista) # [1, 12.3, 'abc', 'd']
valor_removido = lista.pop() #removendo o último elemento, que é o 'd'
print(lista, 'valor removido:', valor_removido)
valor_removido = lista.pop(1) # removendo o valor 12.3 que é o indice 1
print(lista, f'{valor_removido=}')
```

```python
# método clear - limpar a lista

lista = [1, 12.3, 'abc']
print(lista)
lista.clear()
print(lista)
```

```python
# método insert - insere um índice num lugar determinado

lista = [1, 12.3, 'abc']
print(f'{lista[0]=}')
lista.insert(0, 0.5)
print('após o insert de 0.5 no índice 0 ....')
print(f'{lista[0]=}')
print(lista)
```

```python
# método extend e operador + -> extender e concatenar uma ou mais listas

lista_a = [1,2,3]
lista_b = [4,5,6]
lista_a.extend(lista_b)
print(lista_a)

# concatenação de listas
lista_a = [1,2,3]
lista_b = [4,5,6]
lista_c = lista_a + lista_b
print(lista_c)

# lembrando que o método extend não retorna nada, logo ele só deve ser usado num objeto, uma vez que
# ele não é uma função, como del por exemplo.

```

```python
# método remove - remove elemento pelo valor e não pelo indice como o pop e del

# Lista inicial
lista = [1, 2, 3, 4, 2, 5]

# Valor a ser removido
valor_para_remover = 2

# Verificar se o valor está na lista antes de remover
if valor_para_remover in lista:
    #while valor_para_remover in lista:
        lista.remove(valor_para_remover)


print(lista)  # Output: [1, 3, 4, 2, 5]
```

# MÉTODOS INTERESSANTES PARA USARMOS COM LISTAS E STRINGS

Abaixo veremos mais alguns métodos de uso com listas e strings:

```python
# método split - separa uma string por padrão em espaços em branco ou quebras de linha ou caracteres de retorno de carro, veja:
frase = 'python é muito interessante'
lista_palavras = frase.split() #por padrão ele dividirá a string nos espaços
print(lista_palavras)
```

```python
# método join: é um método para unir uma string ou qualquer outro iterável.
# entre as aspas passamos o separador
letras = 'abc'
frase = ''.join(letras)
print(frase)

lista = ['a', 'b', 'c']
frase = ' '.join(lista)
print(frase)

frase = '|'.join(['0','1','2','3'])
print(frase, type(frase))

frase = '/'.join(('0','1','2'))
print(frase, type(frase))

lista = ['a', 'b', 'c']
frase = '-'.join(lista[2])
print(frase)
```
