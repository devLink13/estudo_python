CUIDADOS AO COPIAR LISTAS

No python ao atribuir “cópias” em listas com o operador de igualdade, ‘=’, isso gera uma coisa chamada de cópia raza, ou seja, apenas estamos criando um ponteiro que aponta para o mesmo valor de memória que contém a lista inicial.

```python
lista_a = [1,2,3]
lista_b = lista_a

lista_b[0] = 4
print(lista_a, lista_b) # 4,2,3 e 4,2,3

#perceba que ao alterar um elemento na lista b também estamos alterando na lista a,
# isso acontece porque fizemos um ponteiro apontando para a lista a e não uma cópia
# dos valores da lista  a
```

COMO RESOLVER ISSO?

```python
# PRIMEIRA FORMA USANDO UMA ESPECIE DE 'FATIAMENTO'

lista_a = [1,2,3]
lista_b = lista_a[:]

lista_b[0] = 4
print(f'{lista_a=}', f'{lista_b=}')


# segunda forma é usando o método copy
lista_a = [1,2,3]
lista_b = lista_a.copy()
lista_b[0] = 'amor meu'
print(lista_a)
print(lista_b)
```
