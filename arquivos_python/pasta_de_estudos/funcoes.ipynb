2- DOCSTRING
    
    todas funções em python possui um docstring, que é um manual que será exibido sempre que o usuário solicitar um help da função

```python
#criando uma função contador para criar um docstring dela
# docstrings começam exatamente na linha após o comando def
# apenas precisamos abrir tres aspas duplas depois de def
# veja

def contador (i, f, p):
    """
    TOodo o docstring fica dentro destas aspas
    --> faz uma contagem e mostra na tela
    :param i: inicio da contagem
    :param f: fim da contagem
    :param p: passo da contagem
    :return: sem retorno
    """
    c = i
    while c <= f:
        print(f'{c}', end='...')
        c+=p
    print('FIM!')

help(contador)
```

3- PARAMETROS OPCIONAIS PARA FUNÇÕES:

    são parâmetros que podem ser usados mas não necessariamente é preciso passá-lo

```python
# se fizermos uma função com 3 parâmetros obrigatórios, sempre será necessário passar 3 parâmetros
# se fizermos c=0, isto tornará o parâmetro c como parâmetro opcional.
def somar (a,b,c=0):
    s = a+b+c
    return s

# 3 parâmetros passados
somar(2,3,4)

# 2 parâmetros apenas
somar(2,2)
```

```python
# é possível também passar todos parâmetros como opicionais
# veja:

def somar (a=0, b=0, c=0):
    s = a+b+c
    return s

somar()
```

4- ESCOPO DE VARIÁVEIS

    Refere-se ao lugar onde a variavel em questão pode ser utilizada, o escopo pode ser global ou local.

```python
# escopo da variável teste
def teste():
    x = 8
    print(f'na função teste x vale {x}')
    print(f'na função teste n vale {n}')

# programa principal
n = 2
print(f'no programa principal n vale {n}')

# aqui retornará erro, pois x é uma variável local.
print(f'no programa principal x vale {x}')
```

```python
def teste(b):
    # b e c são variváeis locais
    b+=4
    c = 2

    # OBS: SE EU DEFINIR O "a" DENTRO DE TESTE() ele não será o mesmo 'a' do programa principal
    # ele será o 'a' local da função teste, sendo assim: em python podemos ter variáveis de mesmo nome e valores diferentes
    # desde que estejam em escopos diferentes.
    a = 8
    print(f'"a" dentro vale {a}.')
    print(f'"b" dentro vale {b}.')
    print(f'"C" dentro vale {c}')

# a é uma variável global
# b e c não podem ser usadas aqui pois são globais
a = 5
teste(a)
print(f'"a" fora vale {a}')     
```

```python
# se eu definir que a variável a será global o programa não criará outro espaço de memória para 'a' e assim o valor de
# 'a' tanto dentro quanto fora serão atualizados

def teste(b):
    # b e c são variváeis locais
    b+=4
    c = 2

    # OBS: SE EU DEFINIR O "a" DENTRO DE TESTE() ele não será o mesmo 'a' do programa principal
    # ele será o 'a' local da função teste, sendo assim: em python podemos ter variáveis de mesmo nome e valores diferentes
    # desde que estejam em escopos diferentes.
    global a 
    a = 8
    print(f'"a" dentro vale {a}.')
    print(f'"b" dentro vale {b}.')
    print(f'"C" dentro vale {c}')

# a é uma variável global
# b e c não podem ser usadas aqui pois são globais
a = 5
teste(a)
print(f'"a" fora vale {a}')     
```

5- RETORNO DE VALORES

    Retorno de valores são basicamente fornecer um return para o a função chamada.
    Nem sempre é necessário a função fornecer um return.

```python
def somar (a=0,b=0,c=0):
    s=a+b+c
    return s

r1 = somar(2,5)
r2 = somar(4,9,6)
r3 = somar(2)

print(f'os resultados foram: {r1}, {r2}, {r3}.')
```

6- EXEMPLOS PRÁTICOS


```python
#faça um programa que calcule o fatorial de um numero qualquer e retorne o valor

def fatorial (num = 1):
    f = 1
    for c in range(num, 0, -1):
        f = f * c
    return f    

print(fatorial(int(input('digite um numero para calculo do fatorial: '))))
```

```python
# faça uma função que verifique se um numero é par ou não e retorne verdadeiro para par e falso para impar

def testa_par(num):
    if num % 2 == 0:
        return True
    else:
        return False
    
print(testa_par(3))
print(testa_par(2))
```
